# Advanced Plotting Seaweed Algebra Demo

Be sure to install the seaweed-algebra package!
See the [README](https://github.com/DAflamingFOX/seaweed_algebra/blob/main/README.md#Installation) for more info.

```python
from matplotlib import pyplot as plt

from seaweed_algebra import Meander, Seaweed
from seaweed_algebra.visual import plot_meander, plot_seaweed

# Create a seaweed object, this is in the form of ([a1, a2, ...], [b1, b2, b...])
sw = Seaweed([3, 4], [1, 2, 3, 1])
# Create a meander object, this will copy the same data from the seaweed passed to it.
m = Meander.from_seaweed(sw)

# There are a handful of advanced options which may generate nicer plots

# For seaweed:
# 'nonzero_element' - This can be either '.' or '*' which will show a dot or asterisk for non-zero elements.
# 'draw_diagonal' - Whether or not to show the dotted diagonal line.
# 'draw_meander' - Whether or not to draw the internal meander representation.
# 'color_cycles' - If drawing the meander, whether or not to automatically color the plot.
plot_seaweed(sw, nonzero_element='.', draw_diagonal=False, draw_meander=True, color_cycles=True)

# For meander:
# 'arc_scale' - How much to squish the arcs.
# 'color_cycles' - Whether or not to automatically color the plot.
plot_meander(m, arc_scale=0.5, color_cycles=True)

plt.show()
```

```python
# Additionally, you can export the plots for use in LaTeX documents.
from functools import partial

from seaweed_algebra.latex_utils import export_as_pgf

# Use this to export the plot as a .pgf file, this file
# can then be used in LaTeX using the pgf package.
export_as_pgf(partial(plot_seaweed, sw, draw_meander=True, color_cycles=True), './seaweed.pgf')
```
