```python
from matplotlib import pyplot as plt

from seaweed_algebra import Meander
from seaweed_algebra.visual import plot_meander

# Create a meander object, this is in the form of ([a1, a2, ...], [b1, b2, b...])
m = Meander([17, 3], [10, 4, 6])

# The signature can computed as such, this returns a list
# of "Move" objects which represent the different operations performed.
signature = m.signature()

# We can print out the signature like so:
print(f"The signature of {m} is: {signature}")

# The homotopy can be calculated as such, and returns a list of the component deletions
# Importantly, it returns them in the order that they appear in the signature.
homotopy = m.homotopy()

# The homotopy can be printed out like so:
print(f"The homotopy of {m} is: {homotopy}")

# The component meander can also be calculated:
component = m.component()

# This will display the original and component meander.
fig, (ax1, ax2) = plt.subplots(1, 2)
plot_meander(m, ax1, color_cycles=True)
plot_meander(component, ax2, color_cycles=True)
plt.show()
```
