# Colored Plots Seaweed Algebra Demo

Be sure to install the seaweed-algebra package!
See the [README](https://github.com/DAflamingFOX/seaweed_algebra/blob/main/README.md#Installation) for more info.

```python
from matplotlib import pyplot as plt

from seaweed_algebra import Meander, Seaweed
from seaweed_algebra.visual import plot_meander, plot_seaweed

# Create a seaweed object, this is in the form of ([a1, a2, ...], [b1, b2, b...])
sw = Seaweed([7, 10, 6, 5], [7, 3, 4, 3, 1, 2, 2, 2, 1, 3])
# Create a meander object, this will copy the same data from the seaweed passed to it.
m = Meander.from_seaweed(sw)

# Using matplotlib, plot the seaweed and meander.
# Use the 'color_cycles' parameter to automatically create colored plots.
plot_meander(m, color_cycles=True)
# You can also pass the 'draw_meander' parameter when drawing a seaweed
# to show the meander's path within the seaweed.
plot_seaweed(sw, color_cycles=True, draw_meander=True)

# Show the two plots.
plt.show()
```
